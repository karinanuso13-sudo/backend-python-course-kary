# objetos

```python
class punk_band:
    def __init__(self, band, album, year):
        self.band = band
        self.album = album
        self.year = year
        print ("Una instancia de la clase se ha creado exitosamente")

    def print_attributes(self):
        print ("banda: ", self.band)
        print ("album: ", self.album)
        print ("year: ", self.year)
```

```python
punk_band1 = punk_band ("Blink 182", "One more time", "2023")
```

```python
punk_band1.print_attributes()
```
