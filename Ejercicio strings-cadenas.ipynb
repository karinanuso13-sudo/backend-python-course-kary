```python

my_score =int(input ("Selecciona un score de 1 al 10:"))
positive_words = ["Bueno", "Excelente", "maravilloso", "de calidad"]
negative_words =["Defectuoso", "malo", "horrible", "baja calidad"]
neutral_words = ["aceptable", "regular", "promedio"]
score = int( my_score)

if score > 7:
    words_list = positive_words
    result = "positivo"
elif score == 7:
    words_list = neutral_words
    result = "neutro"
else:
    words_list = negative_words
    result = "negativo"

my_string = "Este es un producto muy " + ", ".join(words_list)

print("analizador de sentimientos\n")
print("my_string : {}\nscore:{} \nresult : {}".format(my_string, score, result))
```
