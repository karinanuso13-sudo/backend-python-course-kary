```python

print("Calculadora (+, -, *, /)")

num1 = float(input("Ingresa un número: "))
operación = input("Ingresa el operador (+, -, *, /): ")
num2 = float(input("Ingresa segundo número: "))

if operación == "+":
    print("Resultado:", num1 + num2)
elif operación == "-":
    print("resultado:", num1 - num2)
elif operación == "*":
    print("resultado:", num1 * num2)
elif operación == "/":
    print ("resultado:", num1 /num2)
```
