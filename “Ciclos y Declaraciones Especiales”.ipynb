```python
import time 
def countup(time_total):
    tiempo_actual = 0
    while tiempo_actual <= time_total:
        mins, secs = divmod (tiempo_actual, 60)
        timeformat = '{:02d} : {:02d}' . format (mins, secs)
        print (timeformat, end = '\r')
        time.sleep(1)
        tiempo_actual += 1
    print ("stop")
countup(90)
    
```
