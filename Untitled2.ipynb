```python
class tienda_discos:
    super_attribute = "Atributo de la super clase"
    def __init__(self, band, album, year):
        self.band = band
        self.album = album
        self.year = year
        print ("Una instancia de la super clase se ha creado exitosamente")

    def print_attributes(self):
        print ("banda: ", self.band)
        print ("album: ", self.album)
        print ("year: ", self.year)
        
class collection_tienda(tienda_discos):
    sub_attribute = "Atributo de la sub clase"
    def __init__(self, band, album, year):
        self.band = band
        self.album = album
        self.year = year
        print ("Una instancia de la sub clase se ha creado exitosamente")

    def print_attributes(self):
        print ("banda: ", self.band)
        print ("album: ", self.album)
        print ("year: ", self.year)
    disco = tienda_discos ("Blink 182", "One more time", "2023")
    disco.print_attributes()
```
